```python
import numpy as np
city_array = np.array([
    "Mumbai", "Delhi", "Bengaluru", "Hyderabad", "Chennai",
    "Pune", "Jaipur", "Ahmedabad", "Kolkata", "Lucknow"
])
city_array=city_array.reshape(2,5)
```

```python
print(city_array[1,3])
print(city_array[1,-1])
print(city_array[:,1:4])
print(city_array[1,:]) # For a row
print(city_array[:,4]) # For a column
print(city_array[1,4::-1]) # For a row in reverse order
```
