```python
import numpy as np
```

```python
# ---------- 1D aggregation: each method returns ONE value ----------
sales = np.array([120, 200, 150, 180, 210, 190, 170])
```

```python
# Total sales (adds all values)
sales.sum()
```

```python
# Average sales
mean_sales = sales.mean()
np.round(mean_sales,2)
```

```python
# Highest sale
sales.max()
```

```python
# Lowest sale
sales.min()
```

```python
# ---------- 2D aggregation: 3 stores (rows) x 4 quarters (columns) ----------
# axis=0 -> down the rows, one result per column | axis=1 -> across columns, one result per row
sales = np.array([[120, 150, 180, 200],
                  [100, 170, 160, 210],
                  [140, 130, 190, 220]])
```

```python
# Whole array total
sales.sum()
```

```python
# Column-wise sum: total of each quarter across all stores
sales.sum(axis=0)
```

```python
# Row-wise sum: total sales of each store
sales.sum(axis=1)
```

```python
# Column-wise average: average sales for each quarter
sales.mean(axis=0)
```

```python
# Row-wise average: average sales for each store
sales.mean(axis=1)
```

```python
# Highest sale per store (best quarter of each store)
sales.max(axis=1)
```

```python
# Lowest sale per quarter (lowest store sale in each quarter)
sales.min(axis=0)
```

```python
# ---------- Median and Percentiles ----------
sales = np.array([120, 150, 180, 200, 210, 220, 260])
```

```python
# Median: middle value after sorting (half below, half above)
sales.sort()
print(sales)
np.median(sales)
```

```python
# 75th percentile: 75% of values are at or below this number
np.percentile(sales, 75)

# Several percentiles at once (25th, 50th = median, 75th)
np.percentile(sales, [25, 50, 75])
```

```python
# ---------- Array functions ----------
sales = np.array([150, 200, 150, 300, 250, 200, 350, 150])  # has duplicates like 150 and 200
```

```python
# Unique values: distinct values, sorted
np.unique(sales)

# Unique values AND how many times each appears
np.unique(sales, return_counts=True)
```

```python
# Square root of each value (element-wise)
np.sqrt(sales)
```

```python
# ==============================
# MORE USEFUL FUNCTIONS
# ==============================
sales = np.array([120, 200, 150, 180, 210, 190, 170])
```

```python
# Standard deviation: how spread out the values are
sales.std()
```

```python
# Position (index) of the highest and lowest sale
print(sales.argmax(), sales.argmin())
```

```python
# Running total: each value = sum of all values up to that point
sales.cumsum()
```

```python
# Sort values from smallest to largest
# sales.sort()
sorted_sales = np.sort(sales)
print(sorted_sales)
print(sales)
```

```python
# Round each value to 2 decimal places
np.round(np.sqrt(sales), 2)
```

```python
# Absolute value: removes the minus sign (e.g. profit/loss -> size of change)
changes = np.array([-20, 15, -5, 30, -10])
np.abs(changes)
```
