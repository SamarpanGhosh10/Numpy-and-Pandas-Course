```python
import numpy as np
```

```python
# Original 3x3 array
test_array = np.array([[1, 2, 3], 
                       [1, 2, 3], 
                       [1, 2, 3]])
test_array
```

```python
# Scalar: 1 is added to every element
test_array + 1
```

```python
# Row vector (3,): stretched down the rows
test_array + np.array([3, 2, 1])
```

```python
# Column vector (3, 1): stretched across the columns
test_array + np.array([[3], 
                       [2], 
                       [1]])
```

```python
# Row vector + column vector: both stretch to (3, 3)
np.array([1, 2, 3]) + np.array([[2], 
                                [2], 
                                [2]])
```

```python
# Shapes don't match: (3, 3) and (2,) -> ValueError
test_array + np.array([2, 1])
```
