```python
import pandas as pd
```

```python
emails = ["anita.dev@corp.com", "rahul.hr@corp.com", "sales.team@corp.com",
          "meera.data@corp.com", "vikram.hr@corp.in", "priya.ops@corp.com"]

email_series = pd.Series(emails, name="emails")
email_series
```

```python
codes = ["ENG-101", "MKT-205", "SAL-330", "HR-118", "ENG-202", "OPS-404"]

code_series = pd.Series(codes, name="codes")
code_series
```

```python
email_series.str.contains("hr")
```

```python
# a bool series can be used to filter straight away
email_series[email_series.str.contains("hr")]
```

```python
email_series[email_series.str.startswith("anita")]
```

```python
email_series.str.endswith(".com")
```

```python
code_series.str.slice(0, 3)
```

```python
code_series.str.slice(-3)
```

```python
# without expand each row keeps a list of the pieces
code_series.str.split("-")
```

```python
# with expand the pieces become separate columns
code_series.str.split("-", expand=True)
```

```python
email_series.str.split("@", expand=True)
```

```python
email_series.str.replace("corp", "company")
```
