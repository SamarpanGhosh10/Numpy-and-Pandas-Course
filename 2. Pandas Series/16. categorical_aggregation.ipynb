```python
import pandas as pd
```

```python
payments = ["UPI", "Card", "UPI", "Cash", "Wallet", "Card",
            "UPI", "Cash", "UPI", "Card", "UPI"]

payment_mode = pd.Series(payments, name="payment_mode")
payment_mode
```

```python
payment_mode.unique()
```

```python
payment_mode.nunique()
```

```python
payment_mode.value_counts()
```

```python
payment_mode.value_counts(normalize=True)
```

```python
# multiply by 100 to read it as a percentage
(payment_mode.value_counts(normalize=True) * 100).round(2)
```

```python
payment_mode.value_counts(ascending=True)
```

```python
payment_mode.value_counts().sort_index()
```

```python
payment_mode.value_counts()["UPI"]
```

```python
(payment_mode == "UPI").sum()
```

```python
# mode gives the most frequent category
payment_mode.mode()
```

```python
# for text data describe shows count, unique, top and freq
payment_mode.describe()
```
