```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([52.0, 47.5, 61.0, 44.0, 58.5])
department = np.array(["engineering", "marketing", "sales", "hr", "finance"])

salary_series = pd.Series(salary, index=department, name="salary")
salary_series
```

```python
print(salary_series + 2)
print(salary_series.add(2))
```

```python
print(salary_series - 5)
print(salary_series.sub(5))
```

```python
print(salary_series * 1.10)
print(salary_series.mul(1.10))
```

```python
print(salary_series / 2)
print(salary_series.div(2))
```

```python
print(salary_series // 2)
print(salary_series.floordiv(2))
```

```python
print(salary_series % 2)
print(salary_series.mod(2))
```

```python
print(salary_series ** 2)
print(salary_series.pow(2))
```

```python
# arithmetic returns a new series, the original one stays unchanged
salary_series
```
