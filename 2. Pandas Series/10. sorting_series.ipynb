```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing", "intern", "sales", "support"])

salary_series = pd.Series(salary, index=department, name="Salary")
salary_series
```

```python
salary_series.sort_values()
```

```python
salary_series.sort_values(ascending=False)
```

```python
salary_series.sort_index()
```

```python
salary_series.sort_index(ascending=False)
```

```python
# sorting returns a new series, the original one stays unchanged
sorted_series = salary_series.sort_values(ascending=False)
salary_series
```

```python
salary_series.nlargest(2)
```

```python
salary_series.nsmallest(2)
```

```python
print(salary_series)
print("-------")
salary_series.rank(ascending=False)
```

```python
# inplace changes the original series instead of returning a new one
salary_series.sort_values(inplace=True)
salary_series
```

```python
# ValueError: ascending takes True or False, not the text "False"
salary_series.sort_values(ascending="False")
```
