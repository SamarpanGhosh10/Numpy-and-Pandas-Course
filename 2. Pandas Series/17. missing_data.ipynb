```python
import numpy as np
import pandas as pd
```

```python
departments = np.array(["HR", "Sales", "Tech", "Finance", "Marketing",
                        "Support", "Design", "Legal", "Admin", "R&D"])
salaries = np.array([55000, np.nan, 72000, 68000, np.nan,
                     55000, 61000, 58000, np.nan, 72000])

salary_series = pd.Series(salaries, index=departments, name="Salary")
salary_series
```
