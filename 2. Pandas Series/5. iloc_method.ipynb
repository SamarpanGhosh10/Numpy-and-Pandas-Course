```python
import pandas as pd
```

```python
salaries = [45000, 52000, 0, 61000, 58000, 72000]
departments = ["developer", "marketing", "hr", "sales", "finance", "support"]

salary_series = pd.Series(salaries, index=departments, name="Salary")
salary_series
```

```python
print(salary_series.iloc[0])
print(salary_series.iloc[2])
print(salary_series.iloc[-1])
print(salary_series.iloc[1:4])
print(salary_series.iloc[[0, 3, 5]])
```

```python
salary_series
```

```python
print(salary_series.iloc[:3])
print(salary_series.iloc[-3:])
print(salary_series.iloc[::2])
print(salary_series.iloc[::-1]) # Reverse
print(salary_series.iloc[1:5:2])
```

```python
salary_series.iloc[2] = 35000
salary_series
```

```python
print(salary_series.iloc[1])
print(salary_series.loc["marketing"])
```

```python
# IndexError: position 10 does not exist, the series has only 6 items
salary_series.iloc[10]
```

```python
# TypeError: iloc works with positions only, it does not accept index labels
salary_series.iloc["hr"]
```
