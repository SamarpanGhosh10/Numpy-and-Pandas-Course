```python
import pandas as pd
```

```python
salaries = [45000, 52000, 0, 61000, 58000, 72000]
departments = ["developer", "marketing", "hr", "sales", "finance", "support"]

salary_series = pd.Series(salaries, index=departments, name="Salary")
salary_series
```

```python
print(salary_series.loc["developer"])
print(salary_series.loc["hr"])
print(salary_series.loc["support"])
print(salary_series.loc["marketing":"sales"])
print(salary_series.loc[["developer", "sales", "support"]])
```

```python
salary_series
```

```python
print(salary_series.loc[:"hr"])
print(salary_series.loc["sales":])
print(salary_series.loc["developer":"finance":2])
print(salary_series.loc[salary_series > 50000])
print(salary_series.loc[salary_series == 0])
```

```python
salary_series.loc["hr"] = 35000
salary_series
```

```python
print(salary_series.loc["marketing"])
print(salary_series.iloc[1])
```
